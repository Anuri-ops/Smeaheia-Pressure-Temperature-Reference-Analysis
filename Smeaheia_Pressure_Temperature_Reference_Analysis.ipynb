# Smeaheia Pressure–Temperature Reference Analysis

Descriptive analysis of four adapted pressure/temperature reference points from the open Smeaheia Dataset.

**Scope:** descriptive data handling and visualisation only. No machine-learning, storage-capacity, pressure-gradient, or causal injection claim is made.


## Data source

Source: Smeaheia Dataset, CO2DataShare (DOI `10.11582/2021.00012`), credited to Equinor and Gassnova.

The local Excel file is an adapted/extracted four-row table and remains subject to the SMEAHEIA DATASET LICENSE.


```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

DATA_PATH = Path("data") / "smeaheia_pressure_temperature_reference.xlsx"
df = pd.read_excel(DATA_PATH).sort_values("Depth_m").reset_index(drop=True)
df
```

## Data-quality check

The dataset contains only four rows. Pressure-state labels are preserved because the values do not all represent the same condition.


```python
print(df)
print("\nMissing values by column:")
print(df.isna().sum())
print("\nPressure types:")
print(df["Pressure_Type"].value_counts())
```

## Reservoir temperature vs depth

The four listed reservoir-temperature values rise with the listed depths. Because the table contains only four adapted reference points, no calibrated geothermal-gradient model is fitted.


```python
plt.figure(figsize=(7, 4.5))
plt.plot(df["Depth_m"], df["Reservoir_Temperature_C"], marker="o")
plt.xlabel("Depth (m)")
plt.ylabel("Reservoir temperature (°C)")
plt.title("Smeaheia reference reservoir temperature vs depth")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## CO₂ injection temperature vs depth

The listed injection-temperature reference values are retained as source values. This notebook does not infer injection design requirements or CO₂ phase behaviour from four points.


```python
plt.figure(figsize=(7, 4.5))
plt.plot(df["Depth_m"], df["CO2_Injection_Temperature_C"], marker="o")
plt.xlabel("Depth (m)")
plt.ylabel("CO₂ injection temperature (°C)")
plt.title("Smeaheia reference CO₂ injection temperature vs depth")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Pressure reference values vs depth

The pressure sequence is non-monotonic and the rows have different pressure-state labels. It is therefore not valid to describe these four values as a single pressure gradient.


```python
plt.figure(figsize=(7, 4.5))
plt.plot(df["Depth_m"], df["Pressure_bar"], marker="o")

for _, row in df.iterrows():
    plt.annotate(
        str(row["Pressure_Type"]),
        (row["Depth_m"], row["Pressure_bar"]),
        xytext=(5, 6),
        textcoords="offset points",
        fontsize=8,
    )

plt.xlabel("Depth (m)")
plt.ylabel("Pressure (bar)")
plt.title("Smeaheia reference pressure values vs depth")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Interpretation boundary

This small exercise demonstrates transparent data handling and descriptive visualisation.

It does **not** support:
- a predictive pressure or temperature model;
- a train/test machine-learning evaluation;
- a storage-efficiency metric;
- a storage-capacity estimate;
- a calibrated pressure or geothermal gradient;
- causal statements about CO₂ injection performance.
